```python

# ANDINASTORE ANALYTICS
# permisos_usuarios
# Permisos directos adicionales a usuarios 
# Ambiente: DEV

CATALOG = "andinastore_dev"

USER_ADMIN = "analyst.demo@andinastore.local"
USER_ANALYST = "jdcapera@ucundinamarca.edu.co"


print("=" * 75)
print("ANDINASTORE - PERMISOS DIRECTOS A USUARIOS")
print("=" * 75)


# ============================================================
# Usuario administrativo 
#
# Permiso adicional 
# USE CATALOG
# ============================================================

spark.sql(f"""
    GRANT USE CATALOG
    ON CATALOG {CATALOG}
    TO `{USER_ADMIN}`
""")

print(f"[OK] USE CATALOG -> {USER_ADMIN}")

# Usuario Analyst 
# USE CATALOG
# USE SCHEMA
# SELECT


spark.sql(f"""
    GRANT USE CATALOG
    ON CATALOG {CATALOG}
    TO `{USER_ANALYST}`
""")

spark.sql(f"""
    GRANT USE SCHEMA
    ON SCHEMA {CATALOG}.gold
    TO `{USER_ANALYST}`
""")

spark.sql(f"""
    GRANT SELECT
    ON SCHEMA {CATALOG}.gold
    TO `{USER_ANALYST}`
""")

print(f"[OK] Acceso Gold de solo lectura -> {USER_ANALYST}")



# VALIDACION CATALOGO


print("\n" + "=" * 75)
print("VALIDACION - CATALOGO")
print("=" * 75)

spark.sql(f"""
    SHOW GRANTS ON CATALOG {CATALOG}
""").show(truncate=False)

# VALIDACION GOLD


print("\n" + "=" * 75)
print("VALIDACION - GOLD")
print("=" * 75)

spark.sql(f"""
    SHOW GRANTS ON SCHEMA {CATALOG}.gold
""").show(truncate=False)


print("\n" + "=" * 75)
print("PERMISOS DIRECTOS A USUARIOS CONFIGURADOS")
print("=" * 75)
```
